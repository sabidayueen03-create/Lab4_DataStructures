```python

LAST_NAME = "Sabida"
SEED_NUM = 7
FAVORITE_ARTIST = "Vhong Navarro"

print("STUDENT INFORMATION")
print("--------------------")
print("Last Name:", LAST_NAME)
print("Seed Number:", SEED_NUM)
print("Favorite Artist:", FAVORITE_ARTIST)
```

```python

character_positions = [
    22, 8, 15, 14, 7, 14,
    1, 22, 1, 18, 18, 15
]

print("Character Positions:")
print(character_positions)
```

```python

customers = [
    "Sabida_C1",
    "Sabida_C2",
    "Sabida_C3",
    "Sabida_C4",
    "Sabida_C5"
]

products = [
    "NOTEBOOK",
    "PEN",
    "USB",
    "HEADPHONES",
    "CALCULATOR"
]

product_prices = {
    "NOTEBOOK": 50,
    "PEN": 15,
    "USB": 350,
    "HEADPHONES": 750,
    "CALCULATOR": 500
}

print("Customers:")
for customer in customers:
    print(customer)

print("\nProducts and Prices:")
for product, price in product_prices.items():
    print(product, ":", price)
```

```python

def generate_quantity(index):
    position = character_positions[index % len(character_positions)]
    return 1 + ((SEED_NUM + position + index) % 5)

print("Generated Quantities:")
for i in range(10):
    print("Transaction", i + 1, ":", generate_quantity(i))
```

```python

transactions = []

for i in range(10):
    customer = customers[i % len(customers)]
    product = products[i % len(products)]
    quantity = generate_quantity(i)
    price = product_prices[product]
    total = quantity * price

    transaction = {
        "customer": customer,
        "product": product,
        "quantity": quantity,
        "unit_price": price,
        "total": total
    }

    transactions.append(transaction)

# Add a repeated transaction for duplicate detection
transactions.append(transactions[0].copy())

print("Transactions generated:", len(transactions))
```

```python

print("TRANSACTION HISTORY")
print("===================")

for transaction in transactions:
    print(transaction)
```

```python

customer_transactions = {}

for transaction in transactions:
    customer = transaction["customer"]

    if customer not in customer_transactions:
        customer_transactions[customer] = []

    customer_transactions[customer].append(transaction)

print("TRANSACTIONS BY CUSTOMER")
print("========================")

for customer, records in customer_transactions.items():
    print("\n", customer)
    for record in records:
        print(
            record["product"],
            "| Quantity:", record["quantity"],
            "| Total:", record["total"]
        )
```

```python

transaction_keys = []

for transaction in transactions:
    key = (
        transaction["customer"],
        transaction["product"],
        transaction["quantity"],
        transaction["unit_price"],
        transaction["total"]
    )
    transaction_keys.append(key)

distinct_transactions = set(transaction_keys)

print("Total Transactions:", len(transactions))
print("Distinct Transactions:", len(distinct_transactions))
print("Repeated Transactions:", len(transactions) - len(distinct_transactions))

print("\nRepeated Transaction Records:")

seen = set()
duplicates = []

for transaction in transactions:
    key = (
        transaction["customer"],
        transaction["product"],
        transaction["quantity"],
        transaction["unit_price"],
        transaction["total"]
    )

    if key in seen:
        duplicates.append(transaction)
    else:
        seen.add(key)

for duplicate in duplicates:
    print(duplicate)
```

```python

transaction_totals = [t["total"] for t in transactions]

print("TRANSACTION VALUE SUMMARY")
print("=========================")
print("Number of Transactions:", len(transactions))
print("Total Sales:", sum(transaction_totals))
print("Average Transaction Value:", round(sum(transaction_totals) / len(transaction_totals), 2))
print("Lowest Transaction Value:", min(transaction_totals))
print("Highest Transaction Value:", max(transaction_totals))
```

```python

print("CUSTOMER SUMMARY")
print("================")

for customer in customers:
    records = customer_transactions.get(customer, [])
    total_spent = sum(record["total"] for record in records)
    quantity_bought = sum(record["quantity"] for record in records)

    print(customer)
    print("  Number of Transactions:", len(records))
    print("  Total Quantity Bought:", quantity_bought)
    print("  Total Amount Spent:", total_spent)

print("\nPRODUCT SUMMARY")
print("===============")

for product in products:
    records = [
        t for t in transactions
        if t["product"] == product
    ]

    total_quantity = sum(t["quantity"] for t in records)
    total_sales = sum(t["total"] for t in records)

    print(product)
    print("  Transactions:", len(records))
    print("  Quantity Sold:", total_quantity)
    print("  Total Sales:", total_sales)
```

```python

ordered_transactions = sorted(
    transactions,
    key=lambda transaction: transaction["total"],
    reverse=True
)

print("TRANSACTIONS ORDERED BY TOTAL VALUE")
print("===================================")

for rank, transaction in enumerate(ordered_transactions, start=1):
    print(
        rank,
        "| Customer:", transaction["customer"],
        "| Product:", transaction["product"],
        "| Quantity:", transaction["quantity"],
        "| Total:", transaction["total"]
    )
```

```python

print("=" * 40)
print("FINAL TRANSACTION HISTORY REPORT")
print("=" * 40)

print("Student:", LAST_NAME)
print("Seed Number:", SEED_NUM)
print("Favorite Artist:", FAVORITE_ARTIST)

print("\nTransaction Statistics")
print("----------------------")
print("Total Transactions:", len(transactions))
print("Distinct Transactions:", len(distinct_transactions))
print("Repeated Transactions:", len(duplicates))
print("Total Sales:", sum(t["total"] for t in transactions))
print("Average Transaction Value:", round(sum(t["total"] for t in transactions) / len(transactions), 2))

highest_transaction = max(
    transactions,
    key=lambda transaction: transaction["total"]
)

lowest_transaction = min(
    transactions,
    key=lambda transaction: transaction["total"]
)

print("\nHighest Transaction:")
print(highest_transaction)

print("\nLowest Transaction:")
print(lowest_transaction)

print("\nReport completed successfully!")
```
