```python
# CELL 1: STUDENT INFORMATION

LAST_NAME = "Sabida"
SEED_NUM = 7
FAVORITE_ARTIST = "Vhong Navarro"

print("Student Information")
print("Last Name:", LAST_NAME)
print("Seed Number:", SEED_NUM)
print("Favorite Artist:", FAVORITE_ARTIST)
```

```python
# CELL 2: CHARACTER POSITIONS

character_positions = [
    ord(char.upper()) - ord('A') + 1
    for char in FAVORITE_ARTIST
    if char.isalpha()
]

print("Character Positions:", character_positions)
```

```python
# ============================================
# CELL 3: STUDENT NAMES AND SUBJECTS
# ============================================


student_names = [
    "Sabida_S1",
    "Sabida_S2",
    "Sabida_S3",
    "Sabida_S4",
    "Sabida_S5"
]

subjects = [
    "PROGRAMMING",
    "MATHEMATICS",
    "ELECTRONICS",
    "PHYSICS",
    "ENGLISH"
]

print("Student names and subjects are ready!")
print(student_names)
print(subjects)
```

```python
# CELL 4: SCORE GENERATIONS FUNCTIONS
def  generate_score(index):
    position = character_positions[index % len(character_positions)]
    return 50 +  ((SEED_NUM * 3 + position + index * 7) % 51)


def get_performance_category(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 80:
        return "VERY GOOD"
    elif score >= 70:
        return "GOOD"
    elif score >= 60:
        return "PASSED"
    else:
        return "NEEDS IMPROVEMENT"

print("Score generation and performance category functions are ready")

```

```python
# ============================================
# CELL 5: GENERATE PERFORMANCE RECORDS
# ============================================

records = []
index = 0

for student in student_names:
    for subject in subjects:
        score = generate_score(index)

        # Include one incomplete record
        if student == student_names[4] and subject == "ENGLISH":
            score = None

        record = {
            "student": student,
            "subject": subject,
            "score": score,
            "category": get_performance_category(score)
        }

        records.append(record)
        index += 1

# Add one repeated student-subject record
records.append(records[0].copy())
```

```python
# ============================================
# ASSESSMENT DATA 1: GENERATED PERFORMANCE RECORDS
# ============================================

print("=" * 65)
print("Generated Student Performance Records:")
print("=" * 65)

for record in records:
    print(record)

print("\nTotal Records:", len(records))
```

```python
# ============================================
# ASSESSMENT DATA 2: ORGANIZED STUDENT/SUBJECT DATA
# ============================================

organized_data = {}

for record in records:
    student = record["student"]
    subject = record["subject"]

    if student not in organized_data:
        organized_data[student] = {}

    if subject not in organized_data[student]:
        organized_data[student][subject] = []

    organized_data[student][subject].append({
        "score": record["score"],
        "category": record["category"]
    })

print("=" * 65)
print("Organized Student/Subject Data:")
print("=" * 65)

for student, subject_data in organized_data.items():
    print(f"\n{student}:")

    for subject, results in subject_data.items():
        print(f"  {subject}: {results}")
```

```python
# ============================================
# ASSESSMENT DATA 3: REPEATED / INCOMPLETE RECORDS
# ============================================

record_keys = [
    (
        record["student"],
        record["subject"],
        record["score"]
    )
    for record in records
]

distinct_records = set(record_keys)

repeated_records = sorted({
    key for key in record_keys
    if record_keys.count(key) > 1
})

incomplete_records = [
    record for record in records
    if record["score"] is None
]

print("=" * 65)
print("Repeated/Incomplete Records:")
print("=" * 65)

print("\nRepeated Records:")
for record in repeated_records:
    print(record)

print("\nIncomplete Records:")
for record in incomplete_records:
    print(record)

print("\nTotal Records:", len(records))
print("Distinct Records:", len(distinct_records))
print("Repeated Record Types:", len(repeated_records))
print("Incomplete Records:", len(incomplete_records))
```

```python
# ============================================
# ASSESSMENT DATA 4: PERFORMANCE CATEGORIES
# ============================================

category_counts = {}

for record in records:
    category = record["category"]
    category_counts[category] = category_counts.get(category, 0) + 1

print("=" * 65)
print("Performance Categories:")
print("=" * 65)

for category, count in sorted(category_counts.items()):
    print(f"{category}: {count}")
```

```python
# ============================================
# ASSESSMENT DATA 5: STUDENT AND SUBJECT SUMMARIES
# ============================================

student_summaries = {}

for student in student_names:
    scores = [
        record["score"]
        for record in records
        if record["student"] == student
        and record["score"] is not None
    ]

    student_summaries[student] = {
        "valid_scores": len(scores),
        "average": round(sum(scores) / len(scores), 2) if scores else None,
        "highest": max(scores) if scores else None,
        "lowest": min(scores) if scores else None
    }

subject_summaries = {}

for subject in subjects:
    scores = [
        record["score"]
        for record in records
        if record["subject"] == subject
        and record["score"] is not None
    ]

    subject_summaries[subject] = {
        "valid_scores": len(scores),
        "average": round(sum(scores) / len(scores), 2) if scores else None,
        "highest": max(scores) if scores else None,
        "lowest": min(scores) if scores else None
    }

print("=" * 65)
print("Student Summaries:")
print("=" * 65)

for student, summary in student_summaries.items():
    print(student, ":", summary)

print("\n" + "=" * 65)
print("Subject Summaries:")
print("=" * 65)

for subject, summary in subject_summaries.items():
    print(subject, ":", summary)
```

```python
# ============================================
# ASSESSMENT DATA 6: ORDERED RESULTS
# ============================================

ordered_results = sorted(
    records,
    key=lambda record: (
        record["score"] is None,
        -(record["score"] if record["score"] is not None else 0),
        record["student"],
        record["subject"]
    )
)

print("=" * 65)
print("Ordered Results (Highest to Lowest):")
print("=" * 65)

for record in ordered_results:
    print(record)
```

```python
# ============================================
# FINAL OUTPUT: STUDENT PERFORMANCE REPORT
# ============================================

valid_scores = [
    record["score"]
    for record in records
    if record["score"] is not None
]

print("=" * 65)
print("FINAL STUDENT PERFORMANCE REPORT")
print("=" * 65)

print("Last Name:", LAST_NAME)
print("Seed Number:", SEED_NUM)
print("Favorite Artist:", FAVORITE_ARTIST)
print("Character Positions:", character_positions)

print("\nOverall Summary")
print("-" * 65)
print("Total Records:", len(records))
print("Distinct Records:", len(distinct_records))
print("Repeated Record Types:", len(repeated_records))
print("Incomplete Records:", len(incomplete_records))
print("Valid Scores:", len(valid_scores))
print("Total of Valid Scores:", sum(valid_scores))
print("Average Valid Score:", round(sum(valid_scores) / len(valid_scores), 2))
print("Highest Score:", max(valid_scores))
print("Lowest Score:", min(valid_scores))

print("\nStudent Performance Summary")
print("-" * 65)

for student, summary in student_summaries.items():
    print(
        f"Student: {student} | "
        f"Valid Scores: {summary['valid_scores']} | "
        f"Average: {summary['average']} | "
        f"Highest: {summary['highest']} | "
        f"Lowest: {summary['lowest']}"
    )

print("\nSubject Performance Summary")
print("-" * 65)

for subject, summary in subject_summaries.items():
    print(
        f"Subject: {subject} | "
        f"Valid Scores: {summary['valid_scores']} | "
        f"Average: {summary['average']} | "
        f"Highest: {summary['highest']} | "
        f"Lowest: {summary['lowest']}"
    )

print("\nPerformance Category Counts")
print("-" * 65)

for category, count in sorted(category_counts.items()):
    print(f"{category}: {count}")

print("\nFinal Report Complete.")
```
