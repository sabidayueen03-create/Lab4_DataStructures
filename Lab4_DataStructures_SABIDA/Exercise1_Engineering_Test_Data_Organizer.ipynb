```python
# ============================================
# TASK 1: ENGINEERING TEST DATA ORGANIZER
# CELL 1 - STUDENT-SPECIFIC INPUTS
# ============================================

LAST_NAME = "Sabida"
SEED_NUM = 7
FAVORITE_ARTIST = "Vhong Navarro"

print("Student Information")
print("Last Name:", LAST_NAME)
print("Seed Number:", SEED_NUM)
print("Favorite Artist:", FAVORITE_ARTIST)
```

```python
# ============================================
# CELL 2 - CHARACTER POSITIONS
# ============================================

character_positions = [
    ord(char.upper()) - ord('A') + 1
    for char in FAVORITE_ARTIST
    if char.isalpha()
]

print("Character Positions:", character_positions)
```

```python
# ============================================
# CELL 3 - TEST SOURCES AND CATEGORIES
# ============================================

test_sources = [
    f"{LAST_NAME}_T1",
    f"{LAST_NAME}_T2",
    f"{LAST_NAME}_T3",
    f"{LAST_NAME}_T4",
    f"{LAST_NAME}_T5"
]

categories = [
    "TEMPERATURE",
    "VOLTAGE",
    "CURRENT",
    "PRESSURE",
    "SPEED"
]

print("Test Sources:")
for source in test_sources:
    print(source)

print("\nMeasurement Categories:")
for category in categories:
    print(category)
```

```python
# ============================================
# CELL 4 - GENERATE MEASUREMENT VALUES
# ============================================

measurement_values = []

for i in range(10):
    position = character_positions[i % len(character_positions)]

    value = SEED_NUM * (i + 1) + position

    measurement_values.append(value)

print("Generated Measurement Values:")
print(measurement_values)
```

```python
# ============================================
# CELL 5 - GENERATE TEST RECORDS
# ============================================

records = [
    {
        "source": test_sources[0],
        "category": categories[0],
        "value": measurement_values[0]
    },
    {
        "source": test_sources[1],
        "category": categories[1],
        "value": measurement_values[1]
    },
    {
        "source": test_sources[2],
        "category": categories[2],
        "value": measurement_values[2]
    },
    {
        "source": test_sources[3],
        "category": categories[3],
        "value": measurement_values[3]
    },
    {
        "source": test_sources[4],
        "category": categories[4],
        "value": measurement_values[4]
    },
    {
        "source": test_sources[0],
        "category": categories[1],
        "value": measurement_values[5]
    },
    {
        "source": test_sources[1],
        "category": categories[2],
        "value": measurement_values[6]
    },
    {
        "source": test_sources[2],
        "category": categories[3],
        "value": measurement_values[7]
    },
    {
        "source": test_sources[3],
        "category": categories[4],
        "value": measurement_values[8]
    },
    {
        "source": test_sources[4],
        "category": categories[0],
        "value": measurement_values[9]
    }
]

# Required repeated test-source/category record
records.append({
    "source": test_sources[0],
    "category": categories[0],
    "value": measurement_values[0]
})
```

```python
# ============================================
# CELL 6 - ASSESSMENT DATA 1
# GENERATED TEST DATA
# ============================================

print("=" * 60)
print("Generated Test Data:")
print("=" * 60)

for record in records:
    print(record)
```

```python
# ============================================
# CELL 7 - ASSESSMENT DATA 2
# ORGANIZED TEST DATA
# ============================================

organized_data = {}

for record in records:

    source = record["source"]
    category = record["category"]
    value = record["value"]

    if source not in organized_data:
        organized_data[source] = {}

    if category not in organized_data[source]:
        organized_data[source][category] = []

    organized_data[source][category].append(value)

print("=" * 60)
print("Organized Test Data:")
print("=" * 60)

for source, category_data in organized_data.items():

    print(source, ":")

    for category, values in category_data.items():
        print("   ", category, ":", values)
```

```python
# ============================================
# ASSESSMENT DATA 3: REPEATED / DISTINCT DATA
# ============================================

record_keys = [
    (r["source"], r["category"], r["value"])
    for r in records
]

distinct_data = set(record_keys)

repeated_data = sorted({
    key for key in record_keys
    if record_keys.count(key) > 1
})

print("=" * 60)
print("Repeated/Distinct Data:")
print("=" * 60)

print("Repeated Records:")
for record in repeated_data:
    print(record)

print("\nDistinct Records:")
for record in sorted(distinct_data):
    print(record)

print("\nTotal Records:", len(records))
print("Distinct Records:", len(distinct_data))
print("Repeated Records:", len(repeated_data))
```

```python
# ============================================
# ASSESSMENT DATA 4: NUMERICAL AND CATEGORICAL SUMMARIES
# ============================================

values = [record["value"] for record in records]

total_value = sum(values)
average_value = total_value / len(values)
minimum_value = min(values)
maximum_value = max(values)

category_counts = {}

for record in records:
    category = record["category"]
    category_counts[category] = (
        category_counts.get(category, 0) + 1
    )

print("=" * 60)
print("Numerical and Categorical Summaries:")
print("=" * 60)

print("Total Count:", len(values))
print("Total Value:", total_value)
print("Average Value:", round(average_value, 2))
print("Minimum Value:", minimum_value)
print("Maximum Value:", maximum_value)

print("\nCategory Counts:")
for category in categories:
    print(category, ":", category_counts.get(category, 0))
```

```python
# ============================================
# ASSESSMENT DATA 5: PROCESSED RESULTS
# ============================================

processed_results = []

for record in records:
    key = (
        record["source"],
        record["category"],
        record["value"]
    )

    processed_results.append({
        "source": record["source"],
        "category": record["category"],
        "value": record["value"],
        "repeated": key in repeated_data
    })

print("=" * 60)
print("Processed Results:")
print("=" * 60)

for record in processed_results:
    print(record)
```

```python
# ============================================
# ASSESSMENT DATA 6: ORDERED RESULTS
# ============================================

ordered_results = sorted(
    records,
    key=lambda record: record["value"],
    reverse=True
)

print("=" * 60)
print("Ordered Results:")
print("=" * 60)

for record in ordered_results:
    print(record)
```

```python
# ============================================
# ASSESSMENT DATA 7: FINAL SUMMARY
# ============================================

print("=" * 60)
print("Final Organized and Summarized Test Dataset:")
print("=" * 60)

for record in ordered_results:
    print(
        f"Source: {record['source']} "
        f"| Category: {record['category']} "
        f"| Value: {record['value']}"
    )

print("\n" + "=" * 60)
print("CONSOLIDATED SUMMARY")
print("=" * 60)

print("Last Name:", LAST_NAME)
print("Seed Number:", SEED_NUM)
print("Favorite Artist:", FAVORITE_ARTIST)

print("Total Records:", len(records))
print("Distinct Records:", len(distinct_data))
print("Repeated Records:", len(repeated_data))
print("Total Measurement Value:", total_value)
print("Average Measurement Value:", round(average_value, 2))
print("Minimum Measurement Value:", minimum_value)
print("Maximum Measurement Value:", maximum_value)

print("\nCategory Summary:")
for category in categories:
    print(
        f"{category}: "
        f"{category_counts.get(category, 0)} measurement(s)"
    )
```
